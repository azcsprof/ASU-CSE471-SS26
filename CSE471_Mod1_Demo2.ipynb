```python
# ============================================================
# DEMO 2 — BREADTH-FIRST SEARCH (BFS)
# ============================================================

# This demo illustrates:
#
# - state exploration
# - frontier expansion
# - visited tracking
# - shortest-path behavior
# - systematic search
#
# The goal is NOT advanced graph theory.
#
# The goal is making exploration visible.
#
# ------------------------------------------------------------
# BEFORE RUNNING:
#
# Predict:
#
# - Which node will expand first?
# - Which node will expand second?
# - Will the search reach G quickly or slowly?
#
# Pay attention to:
#
# - how the frontier changes over time,
# - how nodes wait inside the frontier,
# - and how the search expands outward layer by layer.
#
# BFS explores possibilities in the order they
# were discovered.
# ============================================================


# ------------------------------------------------------------
# GRAPH REPRESENTATION
# ------------------------------------------------------------

# Each node maps to its neighboring nodes.
#
# Example:
#
# "A": ["B", "C"]
#
# means:
#
# A is connected to B and C.

graph = {

    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []

}


# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Try changing the graph structure.
#
# Example:
#
# Add:
#
# "C": ["F", "G"]
#
# Predict:
#
# - Will BFS reach the goal sooner?
# - Which node expands differently?


# ------------------------------------------------------------
# STARTING CONDITIONS
# ------------------------------------------------------------

start_node = "A"
goal_node = "G"


# ------------------------------------------------------------
# FRONTIER
# ------------------------------------------------------------

# BFS uses a queue.
#
# First in → first out.
#
# Nodes discovered earlier
# are explored earlier.

frontier = [start_node]

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# The frontier represents:
#
# "possible future exploration"
#
# Nodes inside the frontier have been discovered,
# but not yet explored.


# ------------------------------------------------------------
# VISITED SET
# ------------------------------------------------------------

# Prevents revisiting nodes.

visited = []

# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Try removing visited tracking later.
#
# Predict:
#
# - What happens if nodes can be revisited forever?
# - Why might cycles become dangerous?


# ------------------------------------------------------------
# STEP COUNTER
# ------------------------------------------------------------

step = 0


# ============================================================
# MAIN SEARCH LOOP
# ============================================================

while len(frontier) > 0:

    print("================================================")
    print("STEP:", step)
    print("================================================")
    print()


    # --------------------------------------------------------
    # SHOW CURRENT FRONTIER
    # --------------------------------------------------------

    print("Current frontier:")
    print(frontier)
    print()

    # --------------------------------------------------------
    # BEFORE CONTINUING:
    #
    # Predict:
    #
    # Which node will be expanded next?
    #
    # Why?
    # --------------------------------------------------------


    # --------------------------------------------------------
    # REMOVE FIRST NODE FROM FRONTIER
    # --------------------------------------------------------

    current = frontier.pop(0)

    print("Expanding node:", current)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # BFS always removes the OLDEST node
    # from the frontier.
    #
    # This creates layer-by-layer exploration.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # GOAL TEST
    # --------------------------------------------------------

    if current == goal_node:

        print("Goal found.")
        break


    # --------------------------------------------------------
    # MARK NODE AS VISITED
    # --------------------------------------------------------

    visited.append(current)

    print("Visited nodes:")
    print(visited)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # Visited nodes have already been explored.
    #
    # The search will not expand them again.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # EXPLORE NEIGHBORS
    # --------------------------------------------------------

    neighbors = graph[current]

    print("Neighbors discovered:")
    print(neighbors)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # Expansion reveals NEW possibilities.
    #
    # Search algorithms repeatedly:
    #
    # - expand states,
    # - discover possibilities,
    # - and organize future exploration.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # ADD UNVISITED NEIGHBORS
    # --------------------------------------------------------

    for neighbor in neighbors:

        # Only add nodes not already explored
        # and not already waiting in the frontier.

        if neighbor not in visited and neighbor not in frontier:

            print("Adding to frontier:", neighbor)

            frontier.append(neighbor)

    print()


    # --------------------------------------------------------
    # SHOW UPDATED FRONTIER
    # --------------------------------------------------------

    print("Updated frontier:")
    print(frontier)
    print()

    # --------------------------------------------------------
    # INTERACTION IDEA:
    #
    # Pause here and inspect the frontier.
    #
    # Ask:
    #
    # - Which node has waited the longest?
    # - Which node will expand next?
    # - How does queue ordering shape behavior?
    # --------------------------------------------------------


    step += 1


# ============================================================
# SEARCH COMPLETE
# ============================================================

print("Search finished.")

# ============================================================
# AFTER RUNNING:
#
# Reflection Questions:
#
# 1. Why does BFS find shallow solutions first?
#
# 2. What role does the frontier play?
#
# 3. Why is visited tracking important?
#
# 4. How does queue ordering shape search behavior?
#
# 5. Why does the search appear systematic?
#
# 6. What would happen in a very large graph?
# ============================================================
```
