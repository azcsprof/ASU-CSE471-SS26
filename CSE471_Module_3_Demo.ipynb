```python
# ============================================================
# MODULE 3 DEMO — GREEDY BEST-FIRST SEARCH
# ============================================================

# This demo illustrates:
#
# - heuristic reasoning
# - guided exploration
# - frontier evaluation
# - selective expansion
# - search prioritization
#
# The goal is NOT advanced optimization.
#
# The goal is understanding how evaluation
# changes search behavior.
#
# ------------------------------------------------------------
# BEFORE RUNNING:
#
# Predict:
#
# - Which node will be explored first?
# - Will the search always choose the shortest path?
# - Which node currently looks "most promising"?
#
# Pay attention to:
#
# - how heuristic values shape exploration,
# - how the frontier gets reordered,
# - and how evaluation changes system behavior.
#
# Unlike BFS:
#
# this search does NOT simply expand
# the oldest node first.
# ============================================================


# ------------------------------------------------------------
# GRAPH STRUCTURE
# ------------------------------------------------------------

graph = {

    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []

}

# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Try modifying the graph.
#
# Example:
#
# Add:
#
# "C": ["F", "G"]
#
# Predict:
#
# - Will the search reach G faster?
# - Which nodes might now be skipped?


# ------------------------------------------------------------
# HEURISTIC VALUES
# ------------------------------------------------------------

# Lower values mean:
#
# "closer to the goal"
#
# The heuristic is an ESTIMATE.
#
# The system uses these estimates
# to guide exploration.

heuristic = {

    "A": 5,
    "B": 4,
    "C": 1,
    "D": 6,
    "E": 2,
    "F": 7,
    "G": 0

}

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# The heuristic does NOT guarantee correctness.
#
# It only provides guidance.
#
# Intelligent systems often rely on:
#
# - approximations,
# - estimates,
# - and incomplete information.


# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Try changing heuristic values.
#
# Example:
#
# Change:
#
# "C": 1
#
# to:
#
# "C": 9
#
# Predict:
#
# - How will exploration behavior change?
# - Which node will now be prioritized?


# ------------------------------------------------------------
# START + GOAL
# ------------------------------------------------------------

start_node = "A"
goal_node = "G"


# ------------------------------------------------------------
# FRONTIER
# ------------------------------------------------------------

# Each entry stores:
#
# (heuristic_value, node)

frontier = [

    (heuristic[start_node], start_node)

]

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# The frontier still represents:
#
# "possible future exploration"
#
# But now the frontier is ORGANIZED
# by evaluation.
#
# The system prioritizes nodes
# that appear more promising.


# ------------------------------------------------------------
# VISITED NODES
# ------------------------------------------------------------

visited = []

# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Later:
#
# compare this demo with BFS.
#
# Ask:
#
# - What changed computationally?
# - What stayed the same?
#
# Both systems explore possibilities,
# but they organize exploration differently.


# ------------------------------------------------------------
# STEP COUNTER
# ------------------------------------------------------------

step = 0


# ============================================================
# MAIN SEARCH LOOP
# ============================================================

while len(frontier) > 0:

    print("================================================")
    print("STEP:", step)
    print("================================================")
    print()


    # --------------------------------------------------------
    # SORT FRONTIER BY HEURISTIC VALUE
    # --------------------------------------------------------

    frontier.sort()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # Sorting reorganizes the frontier.
    #
    # Nodes with smaller heuristic values
    # move toward the front.
    #
    # This changes:
    #
    # - exploration order,
    # - attention,
    # - and search behavior.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # SHOW FRONTIER
    # --------------------------------------------------------

    print("Current frontier:")
    print(frontier)
    print()

    # --------------------------------------------------------
    # BEFORE CONTINUING:
    #
    # Predict:
    #
    # Which node will be selected next?
    #
    # Why?
    # --------------------------------------------------------


    # --------------------------------------------------------
    # SELECT BEST NODE
    # --------------------------------------------------------

    current_h, current_node = frontier.pop(0)

    print("Selected node:", current_node)
    print("Heuristic value:", current_h)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # The search is choosing the node
    # that currently appears closest
    # to the goal.
    #
    # This is a locally greedy strategy.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # GOAL TEST
    # --------------------------------------------------------

    if current_node == goal_node:

        print("Goal found.")
        break


    # --------------------------------------------------------
    # MARK VISITED
    # --------------------------------------------------------

    visited.append(current_node)

    print("Visited nodes:")
    print(visited)
    print()


    # --------------------------------------------------------
    # EXPLORE NEIGHBORS
    # --------------------------------------------------------

    neighbors = graph[current_node]

    print("Neighbors:")
    print(neighbors)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # Expansion reveals NEW possibilities.
    #
    # But unlike BFS:
    #
    # not all possibilities are treated equally.
    # --------------------------------------------------------


    # --------------------------------------------------------
    # EVALUATE EACH NEIGHBOR
    # --------------------------------------------------------

    for neighbor in neighbors:

        if neighbor not in visited:

            h_value = heuristic[neighbor]

            print("Considering:", neighbor)
            print("Heuristic:", h_value)

            frontier.append(

                (h_value, neighbor)

            )

            print("Added to frontier.")
            print()

    # --------------------------------------------------------
    # SHOW UPDATED FRONTIER
    # --------------------------------------------------------

    print("Updated frontier:")
    print(frontier)
    print()

    # --------------------------------------------------------
    # INTERACTION IDEA
    # --------------------------------------------------------

    # Pause and inspect the frontier.
    #
    # Ask:
    #
    # - Which node now looks most promising?
    # - Which node may wait a long time?
    # - How is this different from BFS?
    # --------------------------------------------------------


    step += 1


# ============================================================
# SEARCH COMPLETE
# ============================================================

print("Search finished.")

# ============================================================
# AFTER RUNNING:
#
# Reflection Questions:
#
# 1. Why did the search prioritize certain nodes?
#
# 2. How did heuristic values influence behavior?
#
# 3. Why might heuristics speed up search?
#
# 4. Why might heuristics also mislead systems?
#
# 5. How is this different from BFS?
#
# 6. Why does the system appear more "intelligent"
#    than uninformed search?
# ============================================================
```
