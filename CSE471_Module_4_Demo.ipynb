```python
# ============================================================
# DEMO 4 — CONSTRAINT PROPAGATION
# ============================================================

# This demo illustrates:
#
# - variables
# - domains
# - constraints
# - elimination
# - propagation
#
# The goal is NOT advanced CSP solving.
#
# The goal is making domain reduction visible.
#
# ------------------------------------------------------------
# BEFORE RUNNING:
#
# Predict:
#
# - What happens after assigning A = 1?
# - Which domains will shrink?
# - Will changes spread through the system?
#
# Pay attention to:
#
# - how information propagates,
# - how possibilities disappear,
# - and how constraints reshape future options.
#
# Constraint systems often become powerful
# because they REMOVE impossible states early.
# ============================================================


# ------------------------------------------------------------
# VARIABLES + DOMAINS
# ------------------------------------------------------------

# Each variable starts with
# multiple possible values.
#
# Example:
#
# A: [1, 2, 3]
#
# means:
#
# A could currently be:
# 1 OR 2 OR 3

domains = {

    "A": [1, 2, 3],
    "B": [1, 2, 3],
    "C": [1, 2, 3]

}

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# Domains represent:
#
# "remaining possible values"
#
# Constraint propagation gradually reduces
# these possibilities.


# ------------------------------------------------------------
# CONSTRAINTS
# ------------------------------------------------------------

# Neighboring variables
# cannot have the same value.
#
# A != B
# B != C

constraints = [

    ("A", "B"),
    ("B", "C")

]

# ------------------------------------------------------------
# INTERACTION IDEA
# ------------------------------------------------------------

# Try adding another constraint:
#
# ("A", "C")
#
# Predict:
#
# - Will propagation become stronger?
# - Will domains shrink faster?
#
# More constraints usually create
# more elimination.


# ------------------------------------------------------------
# INITIAL STATE
# ------------------------------------------------------------

print("================================================")
print("INITIAL DOMAINS")
print("================================================")
print()

for variable in domains:

    print(variable, ":", domains[variable])

print()

# ------------------------------------------------------------
# BEFORE CONTINUING:
#
# Ask:
#
# - Which variables currently have uncertainty?
# - Which variables are still flexible?
# ------------------------------------------------------------


# ------------------------------------------------------------
# ASSIGN A VALUE
# ------------------------------------------------------------

# Suppose we now learn:
#
# A = 1
#
# This reduces uncertainty for A.

domains["A"] = [1]


print("================================================")
print("AFTER ASSIGNING A = 1")
print("================================================")
print()

for variable in domains:

    print(variable, ":", domains[variable])

print()

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# One local assignment can affect
# the entire system.
#
# Constraint systems spread information
# through connected variables.


# ============================================================
# CONSTRAINT PROPAGATION
# ============================================================

changed = True

# ------------------------------------------------------------
# PAY ATTENTION
# ------------------------------------------------------------

# The system repeatedly applies constraints
# until no more changes occur.
#
# This repeated stabilization process
# is common in AI systems.
# ------------------------------------------------------------

while changed:

    changed = False

    print("================================================")
    print("PROPAGATION STEP")
    print("================================================")
    print()


    # --------------------------------------------------------
    # CHECK EACH CONSTRAINT
    # --------------------------------------------------------

    for left, right in constraints:

        print("Checking constraint:", left, "!=", right)


        # ----------------------------------------------------
        # IF LEFT SIDE FIXED TO ONE VALUE
        # ----------------------------------------------------

        if len(domains[left]) == 1:

            fixed_value = domains[left][0]

            print(left, "is fixed to:", fixed_value)


            # Remove that value from right side.

            if fixed_value in domains[right]:

                # Only remove if multiple options remain.

                if len(domains[right]) > 1:

                    print("Removing", fixed_value,
                          "from", right)

                    domains[right].remove(fixed_value)

                    changed = True

                    # ----------------------------------------
                    # PAY ATTENTION:
                    #
                    # The system is eliminating
                    # impossible possibilities.
                    #
                    # This reduces future search.
                    # ----------------------------------------


        print()


    # --------------------------------------------------------
    # SHOW CURRENT DOMAINS
    # --------------------------------------------------------

    print("Current domains:")

    for variable in domains:

        print(variable, ":", domains[variable])

    print()

    # --------------------------------------------------------
    # BEFORE NEXT ITERATION:
    #
    # Predict:
    #
    # - Will another domain shrink?
    # - Has the system stabilized?
    #
    # Ask:
    #
    # - What information is spreading?
    # - Which variables still contain uncertainty?
    # --------------------------------------------------------


# ============================================================
# FINAL RESULT
# ============================================================

print("================================================")
print("FINAL DOMAINS")
print("================================================")
print()

for variable in domains:

    print(variable, ":", domains[variable])

print()

# ============================================================
# AFTER RUNNING:
#
# Reflection Questions:
#
# 1. Why did assigning A affect other variables?
#
# 2. How did constraints reduce uncertainty?
#
# 3. Why can elimination reduce search complexity?
#
# 4. Why does propagation sometimes continue
#    through multiple steps?
#
# 5. Why does the system appear to "reason"
#    even though no explicit search tree exists?
#
# 6. How might this approach help solve
#    large scheduling or planning problems?
# ============================================================
```
