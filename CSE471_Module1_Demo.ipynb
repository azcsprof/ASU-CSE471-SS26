```python
# ============================================================
# MODULE 1 DEMO — RATIONAL AGENT IN A GRID WORLD
# ============================================================

# The environment is a simple 5x5 grid.
#
# S = starting position
# G = goal
# X = obstacle
# . = open space
#
# The agent attempts to move toward the goal.
#
# This demo illustrates:
# - perception
# - reasoning
# - action selection
# - environmental constraints
# - rational behavior
#
# ------------------------------------------------------------
# BEFORE RUNNING:
#
# Predict:
# - Which direction will the agent move first?
# - Will the agent always move directly toward the goal?
# - How might obstacles change behavior?
#
# Pay attention to:
# - how the agent evaluates possibilities,
# - how constraints remove possible actions,
# - and how simple rules can produce apparently
#   intelligent behavior.
# ============================================================


# ------------------------------------------------------------
# ENVIRONMENT REPRESENTATION
# ------------------------------------------------------------

grid = [
    ["S", ".", ".", ".", "."],
    [".", "X", "X", ".", "."],
    [".", ".", ".", ".", "."],
    [".", "X", ".", "X", "."],
    [".", ".", ".", ".", "G"]
]

# ------------------------------------------------------------
# INTERACTION IDEA:
#
# Try modifying the environment:
#
# - Remove an obstacle
# - Add a new obstacle
# - Move the goal
#
# Observe:
# - How does agent behavior change?
# - Does the reasoning process stay the same?
# ------------------------------------------------------------


# ------------------------------------------------------------
# INITIAL AGENT POSITION
# ------------------------------------------------------------

agent_row = 0
agent_col = 0


# ------------------------------------------------------------
# GOAL POSITION
# ------------------------------------------------------------

goal_row = 4
goal_col = 4

# ------------------------------------------------------------
# INTERACTION IDEA:
#
# Change the goal location.
#
# Example:
#
# goal_row = 0
# goal_col = 4
#
# Predict:
# - What path will the agent now prefer?
# ------------------------------------------------------------


# ------------------------------------------------------------
# DISPLAY FUNCTION
# ------------------------------------------------------------

def print_grid():

    # Print the environment row by row.

    for r in range(len(grid)):

        row_output = ""

        for c in range(len(grid[0])):

            # Show the agent separately from the grid.

            if r == agent_row and c == agent_col:
                row_output += "A "

            else:
                row_output += grid[r][c] + " "

        print(row_output)

    print()


# ------------------------------------------------------------
# VALID MOVE CHECK
# ------------------------------------------------------------

def valid_position(row, col):

    # Check grid boundaries.

    if row < 0 or row >= len(grid):
        return False

    if col < 0 or col >= len(grid[0]):
        return False

    # Check obstacle collision.

    if grid[row][col] == "X":
        return False

    return True

# ------------------------------------------------------------
# INTERACTION IDEA:
#
# Comment out obstacle checking below:
#
# if grid[row][col] == "X":
#     return False
#
# Observe:
# - What strange behavior appears?
# - Why does obstacle information matter?
# ------------------------------------------------------------


# ------------------------------------------------------------
# DISTANCE TO GOAL
# ------------------------------------------------------------

def distance_to_goal(row, col):

    # Manhattan distance:
    #
    # How many horizontal + vertical moves
    # remain before reaching the goal?

    return abs(goal_row - row) + abs(goal_col - col)

# ------------------------------------------------------------
# PAY ATTENTION:
#
# The agent is NOT planning very far ahead.
#
# It simply evaluates:
# "Which move seems closest to the goal right now?"
#
# This is a local evaluation strategy.
# ------------------------------------------------------------


# ------------------------------------------------------------
# AGENT REASONING
# ------------------------------------------------------------

def choose_action():

    # The agent considers four possible moves.

    possible_moves = [

        ("UP",    agent_row - 1, agent_col),
        ("DOWN",  agent_row + 1, agent_col),
        ("LEFT",  agent_row, agent_col - 1),
        ("RIGHT", agent_row, agent_col + 1)

    ]

    best_move = None
    best_distance = 999999


    # --------------------------------------------------------
    # EVALUATE EACH POSSIBLE ACTION
    # --------------------------------------------------------

    for action, new_row, new_col in possible_moves:

        # Ignore invalid moves.

        if valid_position(new_row, new_col):

            # Compute distance after move.

            dist = distance_to_goal(new_row, new_col)

            print("Considering:", action)
            print("Resulting position:", (new_row, new_col))
            print("Distance to goal:", dist)
            print()

            # Rational choice:
            #
            # Select move minimizing distance.

            if dist < best_distance:

                best_distance = dist
                best_move = (action, new_row, new_col)

    return best_move


# ------------------------------------------------------------
# MAIN EXECUTION LOOP
# ------------------------------------------------------------

step = 0

while True:

    print("================================================")
    print("STEP:", step)
    print("================================================")
    print()

    print_grid()

    # Goal check.

    if agent_row == goal_row and agent_col == goal_col:

        print("Goal reached.")
        break


    # Agent selects action.

    decision = choose_action()


    # Safety check.

    if decision is None:

        print("No valid moves available.")
        break


    action, new_row, new_col = decision


    print("Chosen action:", action)
    print()

    # --------------------------------------------------------
    # PAY ATTENTION:
    #
    # The chosen action changes the environment state.
    #
    # Future decisions depend on:
    # - current position,
    # - remaining possibilities,
    # - and environmental constraints.
    #
    # AI systems often operate through repeated:
    #
    # perceive → evaluate → act
    # --------------------------------------------------------


    # Update environment state.

    agent_row = new_row
    agent_col = new_col


    step += 1


# ============================================================
# AFTER RUNNING:
#
# Reflection Questions:
#
# 1. Did the agent always make the "best" move?
#
# 2. What information was the agent using?
#
# 3. What information was the agent ignoring?
#
# 4. Why does the agent appear intelligent even though
#    the reasoning process is simple?
#
# 5. How might this approach fail in more complex
#    environments?
# ============================================================
```
